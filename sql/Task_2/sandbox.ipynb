```python
import pandas as pd
import polars as pl
```

```python
dfs = []

for i in range(1,8):
    path = f'../../data/umg801_dbs_lecture_data/0{i}_30d_schneider_umg801_dbs_lecture.csv'
    df_i = pl.read_csv(path, n_rows=1)
    dfs.append(df_i)
```

```python
def check_csv_consistency(dfs):
    base_cols = set(dfs[0].columns)
    base_schema = dfs[0].schema

    for i, df in enumerate(dfs):
        if set(df.columns) != base_cols:
            print(f"Column name mismatch in CSV {i}")
        if df.schema != base_schema:
            print(f"Schema mismatch in CSV {i}")
        else:
            print(f'df{i} Columns match schema and names')

check_csv_consistency(dfs)
```
